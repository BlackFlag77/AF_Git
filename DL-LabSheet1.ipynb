```python
# Compare Exponential, Uniform and Normal Distributions
import numpy as np
import matplotlib.pyplot as plt

exp_dist = np.random.exponential(scale=1.0, size=100000)

uniform_dist = np.random.rand(100000)

normal_dist = np.random.normal(size=100000)

plt.figure(figsize=(10,6))

plt.hist(exp_dist,
         density=True,
        #  bins=100,
         bins=150,
         histtype='step',
         color='red',
         label='Exponential')

plt.hist(uniform_dist,
         density=True,
        #  bins=100,
         bins=150,
         histtype='step',
         color='blue',
         label='Uniform')

plt.hist(normal_dist,
         density=True,
        #  bins=100,
         bins=150,
         histtype='step',
         color='green',
         label='Normal')

plt.legend()
plt.title("Random Distributions")
plt.xlabel("Value")
plt.ylabel("Density")

# gets the x axis range to 4 - 8
plt.xlim(-4,8)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

X,Y = np.meshgrid(x,y)

Z = X**2 + Y**2

fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X,Y,Z,cmap='viridis')

ax.set_title("Z = X² + Y²")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

Task 2.4:
Histograms in 1×4 Layout

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1,4, figsize=(20,5))

features = ['N_total',
            'N_bulk',
            'N_surface',
            'R_avg']

for i, col in enumerate(features):
    axes[i].hist(new_df[col], bins=30)
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

Task 2.6: Modified Pairplot

```python
import seaborn as sns

g = sns.PairGrid(new_df)

g.map_upper(sns.histplot)

g.map_diag(sns.histplot)

g.map_lower(sns.kdeplot)

sns.pairplot(new_df)

plt.show()
```
